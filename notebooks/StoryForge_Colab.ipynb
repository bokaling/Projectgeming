```python
#@title 1️⃣ Setup: connect Google Drive and install (run once per session)
#@markdown Allow Google Drive access in the popup: your projects and small models are saved there.

import os, subprocess, sys
REPO = "bokaling/Projectgeming"
APP = "/content/storyforge-app"

def _git(*args):
    return subprocess.run(["git", *args], capture_output=True, text=True).returncode == 0

def _url(token=None):
    return f"https://{token + '@' if token else ''}github.com/{REPO}.git"

def get_code(token=None):
    if os.path.isdir(APP + "/.git"):  # already downloaded: update it
        branch = subprocess.run(["git", "-C", APP, "rev-parse", "--abbrev-ref", "HEAD"], capture_output=True, text=True).stdout.strip()
        return _git("-C", APP, "pull", "-q", "--ff-only", _url(token), branch if branch != "HEAD" else "main") or True
    if not _git("clone", "-q", "--depth", "1", _url(token), APP):
        return False
    if not os.path.exists(APP + "/storyforge/__init__.py"):  # code not merged into main yet: use the build branch
        _git("-C", APP, "fetch", "-q", "--depth", "1", _url(token), "claude/nice-noether-0u7lld:claude/nice-noether-0u7lld")
        _git("-C", APP, "checkout", "-q", "claude/nice-noether-0u7lld")
    _git("-C", APP, "remote", "set-url", "origin", _url())  # never keep a token on disk
    return os.path.exists(APP + "/storyforge/__init__.py")

from google.colab import drive
drive.mount("/content/drive")
print("📥 Downloading StoryForge…")
ok = get_code()
if not ok:
    try:
        from google.colab import userdata
        ok = get_code(userdata.get("GITHUB_TOKEN"))
    except Exception:
        ok = False
if not ok:
    raise SystemExit("❌ Couldn't download StoryForge. Your GitHub repository is private. Either make it public "
                     "(GitHub → Settings → General → Danger Zone → Change visibility), or create a free GitHub token "
                     "and add it as a secret named GITHUB_TOKEN (🔑 on the left, Notebook access ON).")
sys.path.insert(0, APP)
from storyforge import install
install.setup()
print("👉 Now run cell 2️⃣")
```

```python
#@title 2️⃣ Start StoryForge (then click the 👉 OPEN STORYFORGE link)
PUBLIC_LINK = False  #@param {type:"boolean"}
#@markdown ☝️ Tick **PUBLIC_LINK** to also get a link that works on your phone while Colab runs on your computer. Don't share that link.
KEEP_BIG_MODELS_ON_DRIVE = False  #@param {type:"boolean"}
#@markdown ☝️ Only if your Google Drive has 40+ GB free: keeps the big AI models on Drive so they don't download every session.
import sys
sys.path.insert(0, "/content/storyforge-app")
import storyforge
storyforge.launch(public_link=PUBLIC_LINK, cache_big_models_on_drive=KEEP_BIG_MODELS_ON_DRIVE)
```
