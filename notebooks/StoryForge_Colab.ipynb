# ✦ StoryForge: low poly 3D explainer videos ("how it works")
**Before you start:** menu **Runtime → Change runtime type** → **T4 GPU** (cheapest, fast enough), or **CPU** (free, no compute units, slower) → Save.

1. Run **1️⃣ Setup** (▶ button). Allow Google Drive access. Takes 1-3 minutes.
2. Run **2️⃣ Start StoryForge** and click the **👉 OPEN STORYFORGE** link. Keep this tab open.
3. In the app: type a topic (e.g. *how cow poop becomes electricity*) → **✍️ Write the script** → **🚀 Make the video**. Or paste your own script. (First time: press **⚡ Quick test**.)

The app shows how long the session has run and about how many compute units it used. **Disconnect when you're done** (Runtime → Disconnect and delete runtime): a GPU keeps using units while it's connected.

```python
#@title 1️⃣ Setup: connect Google Drive and install (run once per session)
BRANCH = "main"  #@param {type:"string"}
#@markdown Leave **main**. (To test a new version before it goes live, type its branch name here.)

import os, subprocess, sys
REPO = "bokaling/Projectgeming"
APP = "/content/storyforge-app"

def _git(*args):
    return subprocess.run(["git", *args], capture_output=True, text=True).returncode == 0

def _url(token=None):
    return f"https://{token + '@' if token else ''}github.com/{REPO}.git"

def get_code(branch, token=None):
    """Download (or update) StoryForge from GitHub. The token is used only in memory, never saved."""
    if os.path.isdir(APP + "/.git"):
        _git("-C", APP, "fetch", "-q", "--depth", "1", _url(token), f"{branch}:refs/remotes/sf/{branch}")
        _git("-C", APP, "checkout", "-q", "-B", branch, f"refs/remotes/sf/{branch}")
    elif not _git("clone", "-q", "--depth", "1", "--branch", branch, _url(token), APP):
        return False
    _git("-C", APP, "remote", "set-url", "origin", _url())
    return os.path.exists(APP + "/storyforge/__init__.py")

from google.colab import drive
drive.mount("/content/drive")
print(f"📥 Downloading StoryForge ({BRANCH})…")
ok = get_code(BRANCH)
if not ok:
    try:
        from google.colab import userdata
        ok = get_code(BRANCH, userdata.get("GITHUB_TOKEN"))
    except Exception:
        ok = False
if not ok:
    raise SystemExit("❌ Couldn't download StoryForge. Check the BRANCH name. If your GitHub repository is private, "
                     "make it public (GitHub → Settings → General → Danger Zone → Change visibility) or add a free "
                     "GitHub token as a secret named GITHUB_TOKEN (🔑 on the left, Notebook access ON).")
sys.path.insert(0, APP)
from storyforge import install
install.setup()
print("👉 Now run cell 2️⃣")
```

```python
#@title 2️⃣ Start StoryForge (then click the 👉 OPEN STORYFORGE link)
PUBLIC_LINK = False  #@param {type:"boolean"}
#@markdown ☝️ Tick **PUBLIC_LINK** to also get a link that works on your phone while Colab runs on your computer. Don't share that link.
import sys
sys.path.insert(0, "/content/storyforge-app")
import storyforge
storyforge.launch(public_link=PUBLIC_LINK)
```
